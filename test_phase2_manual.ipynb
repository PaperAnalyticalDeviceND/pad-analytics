```python
try:
    # Create a mock pipeline to show detection logic
    class MockPipeline:
        def __init__(self, model_id):
            self.model_id = model_id
        
        def _detect_model_type(self):
            neural_network_models = {16, 17, 19, 20}
            pls_models = {18}
            
            if self.model_id in neural_network_models:
                return 'neural_network'
            elif self.model_id in pls_models:
                return 'pls'
            else:
                return 'unknown'
    
    # Test model detection
    test_models = [16, 17, 18, 19, 20, 99]
    
    print("Model ID -> Detected Type:")
    for model_id in test_models:
        pipeline = MockPipeline(model_id)
        model_type = pipeline._detect_model_type()
        print(f"   Model {model_id:2d} -> {model_type}")
        
except Exception as e:
    print(f"❌ Model detection test failed: {e}")

# Test 3: Show feature name patterns
print("\n3. 🏷️ Feature Name Patterns")
print("-" * 50)

```

```python
import pad_analytics as pad

# Load dataset and get first card
dataset = pad.CachedDataset("FHI2020_Stratified_Sampling")
metadata = dataset.load_dataset_metadata()
card_data = metadata.iloc[0].to_dict()

print(f"🔍 Debugging card {card_data['id']} - {card_data['sample_name']}")

# Method A: Original predict function
print("\n=== ORIGINAL PREDICT ===")
actual, predA = pad.predict(card_id=card_data['id'], model_id=16, verbose=True)
print(f"Original result: {predA}")

# Method B: Model adapter - let's examine each step
print("\n=== ADAPTER PREDICT ===")
adapter = pad.ModelAdapter(model_id=16, auto_load=False)

# Check model info
print("1. Model Info:")
model_info = adapter.get_model_info()
print(f"   Model URL: {model_info.get('model_info', {}).get('model_url', 'None')}")

# Load model and check labels
print("2. Loading model...")
loaded = adapter.load_model()
print(f"   Model loaded: {loaded}")

if loaded:
    # Check labels
    print("3. Labels comparison:")
    model_summary = adapter.get_model_summary()
    adapter_labels = model_summary.get('labels', [])
    print(f"   Adapter labels count: {len(adapter_labels)}")
    print(f"   First 5 adapter labels: {adapter_labels[:5]}")

    # Now check what the original predict function would use for labels
    print("4. Original labels (from get_model):")
    try:
        model_df = pad.get_model(16)
        original_labels = model_df.labels.values[0]
        original_labels = list(map(pad.standardize_names, original_labels))
        print(f"   Original labels count: {len(original_labels)}")
        print(f"   First 5 original labels: {original_labels[:5]}")
        print(f"   Labels match: {adapter_labels == original_labels}")
    except Exception as e:
        print(f"   Error getting original labels: {e}")

    # Make prediction
    print("5. Making adapter prediction...")
    predB = adapter.predict_batch([card_data])
    print(f"   Adapter result: {predB[0]}")

    print(f"\n🎯 COMPARISON:")
    print(f"   Original: {predA}")
    print(f"   Adapter:  {predB[0]}")
    print(f"   Equal:    {predA == predB[0]}")
```

```python
# Complete Example:
import pad_analytics as pad

# Create adapter
adapter = pad.ModelAdapter(model_id=16)

# Load dataset and get cards
dataset = pad.CachedDataset("FHI2020_Stratified_Sampling")
metadata = dataset.load_dataset_metadata()

# Convert to list of dictionaries
cards_list = metadata.head(3).to_dict('records')

print(f"Processing {len(cards_list)} cards...")
print(f"First card: {cards_list[0]['id']} - {cards_list[0]['sample_name']}")
print(f"Sec card: {cards_list[1]['id']} - {cards_list[1]['sample_name']}")
print(f"third card: {cards_list[2]['id']} - {cards_list[2]['sample_name']}")

# Batch prediction
results = adapter.predict_batch(cards_list)

print(f"Results: {len(results)} predictions")
for i, result in enumerate(results):
    print(f"Card {i+1}: {result}")
# Expected Output Format:

# - Neural Network (model 16): List of tuples [(drug_name, confidence, energy), ...]
# - PLS (model 18): List of floats [concentration1, concentration2, ...]
```
