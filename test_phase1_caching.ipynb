# PAD Analytics Phase 1 Caching Test

This notebook tests the Phase 1 caching system functionality.
Run each cell to verify the caching system works correctly.

```python
# Setup and imports
import sys
import os
import time

# Add src to path
sys.path.insert(0, 'src')

import pad_analytics as pad

print(f"PAD Analytics version: {pad.__version__}")
print(f"CacheManager available: {hasattr(pad, 'CacheManager')}")
print(f"CachedDataset available: {hasattr(pad, 'CachedDataset')}")
```

```python
# Test CacheManager
cache_mgr = pad.CacheManager()
print(f"Cache directory: {cache_mgr.cache_dir}")

# Get cache statistics
stats = cache_mgr.get_cache_stats()
print(f"\nCache Statistics:")
for key, value in stats.items():
    print(f"  {key}: {value}")
```

```python
# Test CachedDataset creation and metadata loading
print("Creating CachedDataset...")
dataset = pad.CachedDataset("FHI2020_Stratified_Sampling")

print("\nLoading dataset metadata...")
start_time = time.time()
metadata = dataset.load_dataset_metadata()
load_time = time.time() - start_time

print(f"✅ Loaded {len(metadata)} records in {load_time:.3f}s")
print(f"Dataset representation: {dataset}")
```

```python
# Check cache coverage
print("Checking cache coverage...")
coverage = dataset.get_cache_coverage()

print(f"\nCache Coverage Report:")
for key, value in coverage.items():
    print(f"  {key}: {value}")

print(f"\nDataset is offline ready: {dataset.is_offline_ready()}")
```

```python
# Test performance comparison
print("Testing performance comparison...")

# Create fresh dataset
start_time = time.time()
fresh_dataset = pad.CachedDataset("FHI2020_Stratified_Sampling")
fresh_time = time.time() - start_time

# Load from cache
start_time = time.time()
cached_data = dataset.load_dataset_metadata()
cached_time = time.time() - start_time

print(f"\nPerformance Comparison:")
print(f"  Fresh dataset creation: {fresh_time:.3f}s")
print(f"  Cached data load: {cached_time:.3f}s")

if cached_time < fresh_time:
    speedup = fresh_time / cached_time
    print(f"  🚀 Cache speedup: {speedup:.1f}x faster!")
else:
    print(f"  📝 Similar performance (already optimized)")
```
